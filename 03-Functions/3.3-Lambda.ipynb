## Lambda Function is Python

Lambda Function are small anonymous function defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher order function.

```python
# Syntax
lambda arguments: expression
```

```python
addition = lambda x, y: x + y
type(addition)
print(addition(5, 3))  # Output: 8
```

```python
even=lambda x: x % 2 == 0
even(4)  # Output: True
```

```python
addition = lambda x,y,z: x + y + z
print(addition(5, 3, 2))  # Output: 10
```

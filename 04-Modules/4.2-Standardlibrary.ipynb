## Standard Library Overview

Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.


```python
import array
arr=array.array('i', [1, 2, 3, 4, 5])
print(arr)
```

```python
import math
root=math.sqrt(25)
print(root)
```

```python
## Random Module
import random
print(random.randint(1, 10))
print(random.choice(['apple', 'banana', 'cherry','mango']))
```

```python
## file and directory Access
import os
print(os.getcwd())
print(os.listdir())
```

```python
os.mkdir('new_directory')
```

```python
## High level operation on files and collections of files
import shutil
shutil.copyfile('source.txt', 'destination.txt')
```

```python
## Data Serialization
import json

data = {"name": "Vinay", "age": 28}
json_string = json.dumps(data)
print(json_string)
print(type(json_string))

json_data = json.loads(json_string)
print(json_data)
print(type(json_data))
```

```python
## csv

import csv
with open('data.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Vinay', 28, 'New York'])
    writer.writerow(['Alice', 25, 'Los Angeles'])
    writer.writerow(['Bob', 30, 'Chicago'])
    
with open('data.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime,timedelta
now = datetime.now()    
print(now)
tomorrow = now + timedelta(days=1)
print(tomorrow)
```

```python
## time
import time
print("Current time:", time.ctime())
```

```python
## Regular Expressions
import re

pattern = r'\d+'  # Matches one or more digits
text = "There are 123 apples and 456 oranges."
match = re.findall(pattern, text)
print(match)  # Output: ['123', '456']
```

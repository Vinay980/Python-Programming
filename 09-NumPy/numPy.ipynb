#### NumPy (Numerical Python)
It is a python library designed for working with numbers, arrays, matrices, and mathematical operations efficiently.

#### Vector
A vector is a one-dimensional collection of numerical values.

```python
import numpy as np

print(np.__version__)


numbers=np.array([1,2,3,4,5,6,7,7,8])
print(numbers)
print(type(numbers))
```

#### Matrix
A matrix is a two-dimensional arrangement of numbers organized into rows and columns.

```python
matrix=np.array([[1,2,3],[4,5,6]])
print(matrix)
```

```python
## Get Dimensions
print(numbers.ndim)
matrix.ndim
```

```python
## Get shape
print(numbers.shape)
matrix.shape
```

```python
## Get type
numbers.dtype
```

```python
## get size 
print(numbers.size)
print(matrix.size)
```

```python
## zeros

arr=np.zeros(5)
print(arr)
arr1=np.zeros((2,3))
print(arr1)
```

```python
## ones

arr=np.ones(5)
print(arr)
arr1=np.ones((3,4))
print(arr1)
```

```python
## arrange
arr=np.arange(5)
print(arr)
arr1=np.arange(1,10,2)
print(arr1)
```

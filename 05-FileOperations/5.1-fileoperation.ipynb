## File Operation- Read And Write Files

File handling is crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files and binary files.

```python
## read the whole file

with open('example.txt','r') as file:
    content = file.read()
    print(content)
```

```python
## read the whole file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line)
```

```python
## Writing a file(overwrite)

with open('example.txt','w') as file:
    file.write("This is a new line.\n")
    file.write("This will overwrite the existing content.\n")
```

```python
## writing a file(append)
with open('example.txt','a') as file:
    file.write("This line will be appended to the file.\n")
    file.write("Appending another line.\n")
```

```python
## Writing a list of lines to a file
lines = ["First line\n", "Second line\n", "Third line\n"]
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## Binary file operations
# Writing binary data to a file
with open('binary.bin','wb') as file:
    file.write(b'\x00\x01\x02\x03\x04')
```

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exist, its content is truncated(i.e, the fie is overwritten).

```python
## Writing and then reading a file

with open('example.txt','w+') as file:
    file.write("This is a new line.\n")
    file.write("This will overwrite the existing content.\n")
    file.seek(0)  # Move the cursor to the beginning of the file
    content = file.read()
    print(content)
```

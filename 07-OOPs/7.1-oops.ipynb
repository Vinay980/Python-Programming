## Classes and Objects

Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modelling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including instance variables and methods.



```python
## A class is a blue print for creating objects. Attribute, methods

class Cars:
    pass

audi = Cars()
bmw = Cars()
thar = Cars()
print(type(audi))
```

```python
print(audi)
print(bmw)
print(thar)
```

```python
audi.window = 4

print(audi.window)
```

```python
thar = Cars()
thar.windows = 2
print(thar)
print(thar.windows)
```

```python
dir(thar)
```

```python
## instance variables and methods


class Student:
    ## Constructor

    def __init__(self, name, age):
        self.name = name
        self.age = age

## create object
student1 = Student("vinay",27)
print(student1)
print(student1.name)
print(student1.age)

student2 = Student("ram",25)
print(student2)
print(student2.name)
print(student2.age)
```

```python
## define a class with instance methods
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age=age
    
    def topper(self):
        print(f"{self.name} is topper")

student1 = Student("Ram",28)
student1.topper()
```

```python
## Model of a bank

## Define class of bank account

class BankAccount:
    def __init__(self,owner,balance):
        self.owner= owner
        self.balance= balance
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} deposited. Current balance is {self.balance}")
    def withdrawn(self,amount):
        if amount>self.balance:
            print("Insufficient Funds")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. new balance is {self.balance}")
            
    def get_balance(self):
        return self.balance
    
## Create an account
account=BankAccount("Vinay",10000)
print(account.balance)
```

```python
## Call instance method
account.deposit(10000)
```

```python
account.withdrawn(100)
```

```python
account.get_balance()
```

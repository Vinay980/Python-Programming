#### Stemming
Stemming is the process of reducing a word to its root or base-like form by removing prefixes or suffixes.

#### Porter Stemmer

```python
## Classification Problems
## Comments of product is a postive or negative review
## Reviews ----> playing, played, plays

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = ["play", "playing", "played", "plays", "program", "programming", "programmed", "programs", "eat", "eating", "eaten", "eats", "study", "studying", "studied", "studies"]

for word in words:
    print(f"{word} ----> {stemmer.stem(word)}")
```

```python
stemmer.stem("congratulation")
```

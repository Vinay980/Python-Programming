```python
corpus="""Hello Welcome, to My Learning Journey, My name is Vinay Kumar Mahto. I am a Software Developer and I am going to learn GEN AI. I am very excited to learn this new technology and I hope to become proficient in it. I believe that learning GEN AI will open up new opportunities."""

print(corpus)
```

```python
## Tokenization
## Sentence --> paragraphs

from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
documents
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization
## Paragraphs --> Sentences
## Sentences --> Words

from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer

corpus = "I love Python. I am learning NLP."

tokenizer = TreebankWordTokenizer()

tokens = tokenizer.tokenize(corpus)

print(tokens)
```

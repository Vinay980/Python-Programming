#### Lemmatization
Lemmatization is the process of converting a word into its meaningful base or dictionary form, called a lemma.

```python
## Q&A, chatbots, text summarization


from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize("going",pos="v")
```

```python
words = ["going", "gone", "goes", "went", "go", "better", "best", "good", "bad", "worse", "worst","playing", "played", "plays", "program", "programming", "programmed", "programs", "eat", "eating", "eaten", "eats", "study", "studying", "studied", "studies"]
for word in words:
    print(f"{word}->{lemmatizer.lemmatize(word,pos='v')}")
```
